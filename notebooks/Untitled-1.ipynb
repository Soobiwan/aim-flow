```python
import json

json_path = "D:\\github\\steering\\aim-flow\\configs\\t2i_compbench_100_seed13.json"
input_index = 90

x = 10

with open(json_path, "r", encoding="utf-8") as f:
    data = json.load(f)

samples = data["samples"]

next_prompts = [
    sample["prompt"]
    for sample in samples[input_index : input_index + x]
]

for i, prompt in enumerate(next_prompts, start=input_index):
    print(f"{i}: {prompt}")
```
